```python
# 변수들의 분포를 시각화[4]
# enumerate(features, 1)은 리스트의 요소를 인덱스 + 값 쌍으로 반환합니다.

features = ['age', 'High_BP', 'heart_dz', 'avg_glucose', 'bmi','stroke']
plt.figure(figsize=(15, 10))

for i, col in enumerate(features, 1):
    plt.subplot(2, 3, i)
    sns.histplot(data=stroke, x=col)
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# 수치형 변수와 stroke 상관관계를 히트맵으로 보기
# age, High_BP, heart_dz, avg_glucose, bmi, stroke 데이터 피처들이 상관계수가 1로 아주 높다는 것을 알 수 있다.
stroke_corr = stroke[['age', 'High_BP', 'heart_dz',
                      'avg_glucose', 'bmi', 'stroke']].corr()

sns.heatmap(stroke_corr, annot=True, cmap='coolwarm')
```

<뇌졸중 환자의 나이와 성별 비교>

```python
# 뇌졸중 환자만 필터링 (stroke == 1)
# gender을 제외한 이유는 데이터 시각화에서 보이지않는데 오류가 나서 삭제함
stroke_only = stroke[(stroke['stroke'] == 1) & (stroke['gender'] != 'Other')]

sns.barplot(
    x='gender',
    y='age',
    data=stroke_only,
    palette={'Male': 'lightblue', 'Female': 'lightpink'},
    ci=None

)
plt.title("Age of stroke patients by gender")
```

```python
# Unknown은 데이터 시각화에 불필요한 요소라고 생각해서 지움
# 이전에 담배를 피웠던 사람, 담배를 피우는 사람은 뇌졸중에 걸릴 확률이높음

# formerly smoked -> 뇌졸중 발생률 0.079% ex) 1000명중 79명 꼴
# never smoked -> 뇌졸중 발생률 0.047% ex) 1000명중 47명꼴
# smokes -> 뇌졸중 발생률 0.053% ex) 1000명중 53명꼴

filtered_data = stroke[stroke['smoke'] != 'Unknown']

sns.barplot(
    x='smoke',
    y='stroke',
    data=filtered_data,
    palette={'formerly smoked': 'blue',
             'never smoked': 'green',
             'smokes': 'red'},
    ci=None
)

plt.ylabel("stroke(%)")
plt.title("The link between smoking and stroke")
```

**세계보건기구(WHO) 기준 (일반적인 기준)**
- 정상 체중: BMI 18.5~24.9
- 과체중: BMI 25~29.9
- 비만: BMI 30 이상

```python
# 데이터 시각화 결과 평균 포도당 수치(혈당)가 150이상,
# BMI지수가 30 이상(비만)일경우 뇌졸중에 발생률이 높아진다는 것을 알 수 있습니다.

# stroke를 0과 1이 아닌 Yes or NO로 설정
stroke["stroke_status"] = stroke["stroke"].map({0: "No", 1: "Yes"})

# 산점도 그리기 (stroke_status를 기준으로 색상 분리)
sns.scatterplot(
    x="bmi",
    y="avg_glucose",
    hue="stroke_status",
    data=stroke,
    palette={"No": "blue", "Yes": "red"}  # 색상 지정 (선택 사항)
)

# x축 눈금 설정 (10 간격)
plt.xticks(
    ticks=range(int(stroke["bmi"].min()) // 10 * 10,
                int(stroke["bmi"].max()) + 10, 10)
)

# 그래프 제목
plt.title("The Effect of BMI and Glucose Levels on Stroke")
plt.show()
```

심장병과 고혈압이 모두 있는 사람의 stroke 발생률은 약 **0.22 (22%)**로, 둘 다 없는 사람의 **0.03 (3%)**보다 약 7배 가량 높습니다.
이는 단일 요인보다 복합 위험 요인의 조합이 뇌졸중 발생에 훨씬 강력한 영향을 미친다는 것을 보여줍니다.

따라서 심장병과 고혈압은 뇌졸중의 중요한 발병 원인이라는 것을 알 수 있습니다.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# groupby (그룹별 평균을 계산함) 후 pivot(2차원 표를 만듬)
grouped = stroke.groupby(["heart_dz", "High_BP"])["stroke"].mean().reset_index()
pivoted = grouped.pivot(index="heart_dz", columns="High_BP", values="stroke")

# heatmap으로 시각화
#pivoted 데이터를 색으로 표현
#annot=True: 숫자 값을 셀 안에 표시
#cmap="YlOrRd": 노랑-주황-빨강 색상계열 사용
#fmt=".3f": 소수점 셋째 자리까지 표시
sns.heatmap(pivoted, annot=True, cmap="YlOrRd", fmt=".3f")

# title : 심장질환 및 고혈압에 따른 뇌졸중 발생률
plt.title("Stroke rate by heart_disease and High_BP")
plt.xlabel("High_BP")
plt.ylabel("Heart disease")
plt.show()
```
